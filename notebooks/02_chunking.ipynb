```python
import json

with open("../output/RO_NECP_pages.json", encoding="utf-8") as f:
    pages = json.load(f)

CHUNK=200
OVERLAP=40


chunks=[]
for p in pages:
    words=p["text"].split()
    for j, start in enumerate(range(0, len(words), CHUNK - OVERLAP)):
        piece=" ".join(words[start:start+CHUNK])
        chunks.append({
            "id": f"RO_NECP_p{p['page']}_c{j}",
            "doc": p["doc"], "country": p["country"], "page": p["page"],
            "text": piece,
        })

print(len(chunks), "chunks")
print([c["text"] for c in chunks if "118.35" in c["text"]])
```

```python
with open("../output/RO_NECP_chunks.json", "w", encoding="utf-8") as f:
    json.dump(chunks, f, ensure_ascii=False, indent=2)
```
