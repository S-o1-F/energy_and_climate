### clean and save

```python
import json
import fitz

HEADER = "The 2021-2030 Integrated National Energy and Climate Plan"

doc = fitz.open("../data/romania/First_NECP_Romania.pdf")
pages = []

for i, page in enumerate(doc):
    # remove header from every page and collapse extra spaces and newline
    text = page.get_text().replace(HEADER, "")
    text = " ".join(text.split())
    # Skip first and last page
    if len(text.split()) < 20:
        continue
    pages.append({"doc": "First_NECP_Romania.pdf", "country": "RO", "page": i + 1, "text": text})

# Save all pages, no need for re-reading the pdf
with open("../output/RO_NECP_pages.json", "w", encoding="utf-8") as f:
    json.dump(pages, f, ensure_ascii=False, indent=2)

print(len(pages), "pages saved")
```
